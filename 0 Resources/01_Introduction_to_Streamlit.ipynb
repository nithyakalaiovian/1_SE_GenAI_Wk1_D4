# 1. Introduction to Streamlit

**Streamlit** is a free Python library used to build **web apps** quickly — without needing to know HTML, CSS, or JavaScript.

With just simple Python code, you can create interactive websites, dashboards, and data apps.

## Why use Streamlit?

- ✅ **Very easy** — write only Python.
- ✅ **Fast** — build an app in minutes.
- ✅ **Interactive** — add buttons, sliders, charts easily.
- ✅ **Popular** — widely used for Data Science and AI demos.

### Real examples of what you can build
- A calculator app
- A data dashboard with charts
- A machine learning prediction app
- A chatbot interface

## Step 1: Install Streamlit

Type this command in the **terminal** (not inside Python):

```
pip install streamlit
```

> Tip: Do this inside a virtual environment if you have one active.

## Step 2: Write your first app

Streamlit apps are saved as a **`.py`** file (NOT a notebook).

Create a file called **`app.py`** and put this code inside it:

```python
import streamlit as st

st.title("My First Streamlit App")
st.write("Hello! Welcome to Streamlit.")
```

- `import streamlit as st` → loads Streamlit (we call it `st` for short).
- `st.title()` → shows a big heading.
- `st.write()` → shows text (or almost anything).

## Step 3: Run the app

In the terminal, type:

```
streamlit run app.py
```

This opens your app automatically in the **web browser** (usually at `http://localhost:8501`).

To stop the app, press **Ctrl + C** in the terminal.
