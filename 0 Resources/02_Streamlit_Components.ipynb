# 2. Python Streamlit Components

**Components** (also called widgets) are the building blocks of a Streamlit app — like text, buttons, sliders, and inputs.

Remember: put this code in a `.py` file and run with `streamlit run filename.py`.

We import Streamlit once at the top:
```python
import streamlit as st
```

## 1. Text components (showing information)

| Component | What it does |
|-----------|--------------|
| `st.title()` | Big heading |
| `st.header()` | Medium heading |
| `st.subheader()` | Small heading |
| `st.write()` | Normal text (or anything) |
| `st.markdown()` | Text with formatting (**bold**, etc.) |

````python
import streamlit as st

st.title("This is a Title")
st.header("This is a Header")
st.subheader("This is a Subheader")
st.write("This is normal text.")
st.markdown("This is **bold** text using markdown.")
````

## 2. Input components (getting data from the user)

These let the user type or choose something.

````python
import streamlit as st

# Text input
name = st.text_input("Enter your name:")
st.write("Hello,", name)

# Number input
age = st.number_input("Enter your age:")
st.write("Your age is", age)
````

## 3. Button

A button does something **when clicked**.

The code inside the `if` runs only when the button is pressed.

````python
import streamlit as st

if st.button("Click Me"):
    st.write("You clicked the button! 🎉")
````

## 4. Slider

Lets the user pick a number by sliding.

````python
import streamlit as st

# slider(label, min, max)
number = st.slider("Pick a number", 0, 100)
st.write("You picked:", number)
````

## 5. Selectbox and Checkbox

- **Selectbox** → choose one option from a dropdown.
- **Checkbox** → a tick box (True/False).

````python
import streamlit as st

# Dropdown
city = st.selectbox("Choose your city", ["Chennai", "Delhi", "Mumbai"])
st.write("You selected:", city)

# Checkbox
agree = st.checkbox("I agree to the terms")
if agree:
    st.write("Thank you for agreeing!")
````

## 6. Showing messages

Nice colored boxes for feedback.

````python
import streamlit as st

st.success("This is a success message (green).")
st.info("This is an info message (blue).")
st.warning("This is a warning message (yellow).")
st.error("This is an error message (red).")
````

## Key points to remember

- **Text**: `title`, `header`, `subheader`, `write`, `markdown`.
- **Inputs**: `text_input`, `number_input`.
- **Widgets**: `button`, `slider`, `selectbox`, `checkbox`.
- **Messages**: `success`, `info`, `warning`, `error`.
- Widgets return the user's value, which you can store in a variable.
