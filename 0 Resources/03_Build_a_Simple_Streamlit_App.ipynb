# 3. Build a Simple Streamlit App

Now let's combine what we learned and build a **complete mini app**.

We will build a **BMI Calculator** (Body Mass Index) — it takes your weight and height and tells your BMI.

> A ready-to-run file named **`app.py`** is included in this same folder. Run it with:
> ```
> streamlit run app.py
> ```

## Step 1: Plan the app

Our app needs to:
1. Show a **title**.
2. Ask for **weight** (in kg).
3. Ask for **height** (in meters).
4. Have a **button** to calculate.
5. Show the **BMI result**.

## Step 2: The complete code

Copy this code into a file named `app.py` (or use the one already provided), then run `streamlit run app.py`.

````python
import streamlit as st

# 1. Title
st.title("BMI Calculator")
st.write("Enter your details to calculate your BMI.")

# 2. Get inputs from the user
weight = st.number_input("Enter your weight (kg):", min_value=1.0)
height = st.number_input("Enter your height (meters):", min_value=0.1)

# 3. Button to calculate
if st.button("Calculate BMI"):
    bmi = weight / (height * height)
    st.write("Your BMI is:", round(bmi, 2))

    # 4. Give a message based on the result
    if bmi < 18.5:
        st.warning("You are underweight.")
    elif bmi < 25:
        st.success("You have a normal weight. 👍")
    else:
        st.error("You are overweight.")
````

## Step 3: Understanding the code

- `st.title()` → shows the app heading.
- `st.number_input()` → asks the user for numbers.
- `st.button("Calculate BMI")` → the code inside runs only when clicked.
- `bmi = weight / (height * height)` → the BMI formula.
- `if / elif / else` → shows a different message based on the BMI value.
- `st.success / warning / error` → colored result messages.

## Step 4: Run it!

1. Open the terminal in this folder.
2. Type:
   ```
   streamlit run app.py
   ```
3. Your app opens in the browser.
4. Enter weight and height, click **Calculate BMI**, and see the result.
